Este primeiro codígo foca em montar a base da fila dinâmica e validar a regra de negócios. Ele cria as classes Paciente, Node e FilaAtendimento e, no bloco final, já roda uma simulação automatizada. A ideia aqui é provar que a manipulação dos ponteiros está funcionando direitinho — garantindo que os pacientes prioritários avancem na frente dos normais, mas respeitem a ordem de chegada entre si, tudo isso sem você precisar ficar digitando os dados na mão toda vez que for testar a lógica.

```python
class Paciente:
    """O Paciente: Nome, idade, nível de prioridade."""
    def __init__(self, nome, idade, nivel_prioridade):
        self.nome = nome
        self.idade = idade
        self.nivel_prioridade = nivel_prioridade

    def __str__(self):
        return f"[{self.nome}, {self.idade} anos - {self.nivel_prioridade.upper()}]"


class Node:
    """A estrutura básica de um Nó dinâmico contendo o dado e a referência do próximo."""
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    """
    A Dinâmica: Pacientes chegam de forma contínua e saem quando são atendidos.
    """
    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    # ==========================================
    # O Contrato (Operações):
    # ==========================================

    def adicionar(self, paciente):
        """1. Adicionar paciente"""
        novo_no = Node(paciente)

        # CASO 1: Fila está totalmente vazia
        if self.esta_vazia():
            self.inicio = novo_no
            self.fim = novo_no
            print(f"-> Paciente {paciente.nome} adicionado como o primeiro da fila.")

        # CASO 2: Paciente é PRIORITÁRIO
        elif paciente.nivel_prioridade == "Prioridade":
            # O primeiro paciente da fila é Normal -> Novo nó entra no início
            if self.inicio.paciente.nivel_prioridade != "Prioridade":
                novo_no.proximo = self.inicio
                self.inicio = novo_no
            else:
                # Percorre a fila para encontrar o último prioritário
                atual = self.inicio
                while atual.proximo is not None and atual.proximo.paciente.nivel_prioridade == "Prioridade":
                    atual = atual.proximo

                # Reorganiza os ponteiros para encaixar o nó prioritário
                novo_no.proximo = atual.proximo
                atual.proximo = novo_no

                # Atualiza a referência do fim se inserido no final
                if novo_no.proximo is None:
                    self.fim = novo_no
            print(f"-> Paciente {paciente.nome} adicionado à fila (PRIORIDADE).")

        # CASO 3: Paciente NÃO PRIORITÁRIO (Insere sempre no final)
        else:
            self.fim.proximo = novo_no
            self.fim = novo_no
            print(f"-> Paciente {paciente.nome} adicionado à fila.")

        self._tamanho += 1

    def atender(self):
        """2. Atender (remover o primeiro)"""
        if self.esta_vazia():
            print("\nA fila está vazia! Nenhum paciente para atender.")
            return None

        paciente_atendido = self.inicio.paciente

        # Remove o primeiro da fila e move o ponteiro 'inicio'
        self.inicio = self.inicio.proximo
        self._tamanho -= 1

        # Se a fila ficou vazia, o fim também vira None
        if self.inicio is None:
            self.fim = None

        return paciente_atendido

    def listar_espera(self):
        """3. Listar espera"""
        print("\n--- FILA DE ESPERA ATUAL ---")
        if self.esta_vazia():
            print("Nenhum paciente aguardando.")
            print("----------------------------\n")
            return

        atual = self.inicio
        posicao = 1

        while atual is not None:
            print(f"{posicao}º -> {atual.paciente}")
            atual = atual.proximo
            posicao += 1

        print(f"Total na fila: {self._tamanho}")
        print("----------------------------\n")

    def esta_vazia(self):
        """4. Checar se está vazia"""
        return self.inicio is None

    # Função extra para auxiliar no rastreamento de tamanho
    def tamanho(self):
        return self._tamanho


# ==========================================
# TESTES DO SISTEMA
# ==========================================
if __name__ == "__main__":
    fila = FilaAtendimento()

    print("=== 1. CHEGADA DE PACIENTES NORMIAIS E PRIORITÁRIOS ===")

    fila.adicionar(Paciente("Ana", 32, "Normal"))
    fila.adicionar(Paciente("Bruno", 70, "Normal"))
    fila.adicionar(Paciente("Carlos", 45, "Prioridade"))
    fila.adicionar(Paciente("Diana", 65, "Prioridade"))
    fila.adicionar(Paciente("Eduardo", 20, "Normal"))

    # 3. Listar espera
    fila.listar_espera()

    print("=== 2. ATENDIMENTO DOS PACIENTES ===")

    paciente_atendido = fila.atender()
    print(f"✔ Atendendo: {paciente_atendido.nome}")

    paciente_atendido = fila.atender()
    print(f"✔ Atendendo: {paciente_atendido.nome}")

    fila.listar_espera()

    print("=== 3. ATENDENDO O RESTANTE DOS PACIENTES ===")
    print(f"✔ Atendendo: {fila.atender().nome}")
    print(f"✔ Atendendo: {fila.atender().nome}")
    print(f"✔ Atendendo: {fila.atender().nome}")

    # 4. Checar se está vazia na prática
    print(f"A fila está vazia agora? {fila.esta_vazia()}")
    fila.atender()
```

Com a adição de um laço contínuo (while True) e entradas do usuário (input), o script deixa de ser apenas um teste estático e vira um sisteminha funcional. É a melhor opção para simular o fluxo real da clínica na prática, permitindo cadastrar quem chega e chamar para atendimento de forma manual e dinâmica.

```python
class Paciente:
    """O Paciente: Nome, idade, nível de prioridade."""
    def __init__(self, nome, idade, nivel_prioridade):
        self.nome = nome
        self.idade = idade
        self.nivel_prioridade = nivel_prioridade

    def __str__(self):
        return f"[{self.nome}, {self.idade} anos - {self.nivel_prioridade.upper()}]"


class Node:
    """A estrutura básica de um Nó dinâmico contendo o dado e a referência do próximo."""
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    """A Dinâmica: Pacientes chegam de forma contínua e saem quando são atendidos."""
    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def adicionar(self, paciente):
        """1. Adicionar paciente"""
        novo_no = Node(paciente)

        if self.esta_vazia():
            self.inicio = novo_no
            self.fim = novo_no
            print(f"\n-> Paciente {paciente.nome} adicionado como o primeiro da fila.")

        elif paciente.nivel_prioridade == "Prioridade":
            if self.inicio.paciente.nivel_prioridade != "Prioridade":
                novo_no.proximo = self.inicio
                self.inicio = novo_no
            else:
                atual = self.inicio
                while atual.proximo is not None and atual.proximo.paciente.nivel_prioridade == "Prioridade":
                    atual = atual.proximo

                novo_no.proximo = atual.proximo
                atual.proximo = novo_no

                if novo_no.proximo is None:
                    self.fim = novo_no
            print(f"\n-> Paciente {paciente.nome} adicionado à fila (PRIORIDADE).")

        else:
            self.fim.proximo = novo_no
            self.fim = novo_no
            print(f"\n-> Paciente {paciente.nome} adicionado à fila.")

        self._tamanho += 1

    def atender(self):
        """2. Atender (remover o primeiro)"""
        if self.esta_vazia():
            print("\nA fila está vazia! Nenhum paciente para atender.")
            return None

        paciente_atendido = self.inicio.paciente
        self.inicio = self.inicio.proximo
        self._tamanho -= 1

        if self.inicio is None:
            self.fim = None

        return paciente_atendido

    def listar_espera(self):
        """3. Listar espera"""
        print("\n--- FILA DE ESPERA ATUAL ---")
        if self.esta_vazia():
            print("Nenhum paciente aguardando.")
            print("----------------------------")
            return

        atual = self.inicio
        posicao = 1

        while atual is not None:
            print(f"{posicao}º -> {atual.paciente}")
            atual = atual.proximo
            posicao += 1

        print(f"Total na fila: {self._tamanho}")
        print("----------------------------")

    def esta_vazia(self):
        """4. Checar se está vazia"""
        return self.inicio is None

    def tamanho(self):
        return self._tamanho


# ==========================================
# INTERFACE DO USUÁRIO (MENU INTERATIVO)
# ==========================================
def iniciar_sistema():
    fila = FilaAtendimento()

    while True:
        print("\n" + "="*30)
        print("    SISTEMA DA CLÍNICA")
        print("="*30)
        print("1. Adicionar paciente")
        print("2. Atender paciente")
        print("3. Listar espera")
        print("4. Checar se fila está vazia")
        print("0. Sair")
        print("="*30)

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            nome = input("Digite o nome do paciente: ")

            # Validação simples para garantir que a idade é um número
            try:
                idade = int(input("Digite a idade: "))
            except ValueError:
                print("Idade inválida! Tente adicionar novamente.")
                continue

            print("Nível de prioridade:")
            print("  [1] Normal")
            print("  [2] Prioridade")
            op_prioridade = input("Escolha: ")

            nivel = "Prioridade" if op_prioridade == "2" else "Normal"

            # Cria o paciente e joga na fila
            novo_paciente = Paciente(nome, idade, nivel)
            fila.adicionar(novo_paciente)

        elif opcao == "2":
            paciente = fila.atender()
            if paciente is not None:
                print(f"\n✔ ATENDENDO AGORA: {paciente.nome} ({paciente.idade} anos - {paciente.nivel_prioridade})")

        elif opcao == "3":
            fila.listar_espera()

        elif opcao == "4":
            if fila.esta_vazia():
                print("\nSituação: A fila está VAZIA no momento.")
            else:
                print(f"\nSituação: A fila NÃO está vazia. Há {fila.tamanho()} paciente(s) aguardando.")

        elif opcao == "0":
            print("\nEncerrando o sistema da clínica... Até logo!")
            break

        else:
            print("\nOpção inválida! Por favor, escolha um número de 0 a 4.")

# Roda o sistema
if __name__ == "__main__":
    iniciar_sistema()
```
